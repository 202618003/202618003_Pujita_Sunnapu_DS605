## Task 2 - Vectorized Arithmetic and Linear Algebra

```python
# -----------------------------
# Task 2: Vectorized Arithmetic and Linear Algebra
# -----------------------------

# Create two matrices
matrix_A = np.array([
    [1, 2],
    [3, 4]
])

matrix_B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(matrix_A)

print("\nMatrix B:")
print(matrix_B)


# 1. Matrix addition
addition = matrix_A + matrix_B

print("\nMatrix Addition (A + B):")
print(addition)


# 2. Element-wise multiplication
elementwise_multiplication = matrix_A * matrix_B

print("\nElement-wise Multiplication (A * B):")
print(elementwise_multiplication)


# 3. Matrix multiplication
matrix_multiplication = matrix_A @ matrix_B

print("\nMatrix Multiplication (A @ B):")
print(matrix_multiplication)


# 4. Transpose
transpose_A = matrix_A.T

print("\nTranspose of Matrix A:")
print(transpose_A)


# 5. Determinant
determinant_A = np.linalg.det(matrix_A)

print("\nDeterminant of Matrix A:")
print(determinant_A)


# 6. Inverse
if not np.isclose(determinant_A, 0):
    inverse_A = np.linalg.inv(matrix_A)

    print("\nInverse of Matrix A:")
    print(inverse_A)

    # Verify A @ A^-1 = Identity Matrix
    identity_check = matrix_A @ inverse_A

    print("\nA @ A^-1:")
    print(identity_check)

    print("\nIs A @ A^-1 approximately the identity matrix?",
          np.allclose(identity_check, np.eye(matrix_A.shape[0])))
else:
    print("\nMatrix A is singular and does not have an inverse.")
```

## Task 3 - Normal Distribution and Histogram

```python
# -----------------------------
# Task 3: Normal Distribution and Histogram
# -----------------------------

# Parameters of the normal distribution
chosen_mean = 50
chosen_std = 10

# Generate 1,000 values from a normal distribution
normal_data = np.random.normal(
    loc=chosen_mean,
    scale=chosen_std,
    size=1000
)

# Calculate sample mean and sample standard deviation
sample_mean = np.mean(normal_data)
sample_std = np.std(normal_data, ddof=1)

print("Chosen mean:", chosen_mean)
print("Chosen standard deviation:", chosen_std)

print("\nSample mean:", sample_mean)
print("Sample standard deviation:", sample_std)

# Compare chosen values with sample statistics
print("\nDifference between chosen and sample mean:",
      sample_mean - chosen_mean)

print("Difference between chosen and sample standard deviation:",
      sample_std - chosen_std)


# Plot histogram
plt.figure(figsize=(8, 5))

plt.hist(normal_data, bins=30, edgecolor="black")

plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

# Part B - Data Wrangling with Pandas: Titanic Dataset

## Task 4 - Load and Inspect Data

```python
# -----------------------------
# Task 4: Load and Inspect Titanic Dataset
# -----------------------------

# Load the Titanic dataset
titanic_df = pd.read_csv("train.csv")


# 1. Display first 5 rows
print("First 5 rows:")
display(titanic_df.head())


# 2. Display last 5 rows
print("\nLast 5 rows:")
display(titanic_df.tail())


# 3. Display shape
print("\nShape of dataset:")
print(titanic_df.shape)


# 4. Display column names
print("\nColumn names:")
print(titanic_df.columns)


# 5. Display dataset information
print("\nDataset information:")
titanic_df.info()


# 6. Display descriptive statistics
print("\nDescriptive statistics:")
display(titanic_df.describe())


# 7. Example using loc
print("\nExample using loc:")
display(titanic_df.loc[0:4, ["Name", "Sex", "Age"]])


# 8. Example using iloc
print("\nExample using iloc:")
display(titanic_df.iloc[0:5, 3:6])


# 9. Difference between loc and iloc
print("\nDifference between loc and iloc:")
print("loc selects data using row/column labels.")
print("iloc selects data using integer positions.")
```

## Task 5 - Filtering and Querying

```python
# -----------------------------
# Task 5: Filtering and Querying
# -----------------------------

# 1. Male passengers older than 50
male_over_50 = titanic_df[
    (titanic_df["Sex"] == "male") &
    (titanic_df["Age"] > 50)
]

male_over_50_count = len(male_over_50)

print("1. Male passengers older than 50:", male_over_50_count)


# 2. Female first-class passengers and their survival percentage
female_first_class = titanic_df[
    (titanic_df["Sex"] == "female") &
    (titanic_df["Pclass"] == 1)
]

female_first_class_count = len(female_first_class)
female_first_class_survival_rate = female_first_class["Survived"].mean() * 100

print("\n2. Female first-class passengers:", female_first_class_count)
print("   Survival percentage:",
      female_first_class_survival_rate)


# 3. Age 20-40, Fare above overall median, and survived
overall_median_fare = titanic_df["Fare"].median()

age_20_40_high_fare_survived = titanic_df[
    (titanic_df["Age"] >= 20) &
    (titanic_df["Age"] <= 40) &
    (titanic_df["Fare"] > overall_median_fare) &
    (titanic_df["Survived"] == 1)
]

print("\n3. Age 20-40, Fare above overall median, and survived:",
      len(age_20_40_high_fare_survived))

print("   Overall median Fare:", overall_median_fare)


# 4. Travelling alone, age < 30, and did not survive
travelling_alone_under_30_not_survived = titanic_df[
    (titanic_df["SibSp"] == 0) &
    (titanic_df["Parch"] == 0) &
    (titanic_df["Age"] < 30) &
    (titanic_df["Survived"] == 0)
]

print("\n4. Travelling alone, age < 30, and did not survive:",
      len(travelling_alone_under_30_not_survived))


# 5. Embarked='S', Pclass 2 or 3, Fare above Southampton median
southampton_median_fare = titanic_df.loc[
    titanic_df["Embarked"] == "S",
    "Fare"
].median()

southampton_class_2_3_high_fare = titanic_df[
    (titanic_df["Embarked"] == "S") &
    (titanic_df["Pclass"].isin([2, 3])) &
    (titanic_df["Fare"] > southampton_median_fare)
]

print("\n5. Embarked='S', Pclass 2 or 3, and Fare above Southampton median:",
      len(southampton_class_2_3_high_fare))

print("   Southampton median Fare:", southampton_median_fare)
```

## Task 6 - groupby() and Aggregation

```python
# -----------------------------
# Task 6: Groupby() and Aggregation
# -----------------------------

# 1. Survival rate by Sex
survival_by_sex = (
    titanic_df.groupby("Sex")["Survived"]
    .mean()
    .mul(100)
)

print("1. Survival rate by Sex:")
display(survival_by_sex)


# 2. Survival rate by Pclass
survival_by_pclass = (
    titanic_df.groupby("Pclass")["Survived"]
    .mean()
    .mul(100)
)

print("\n2. Survival rate by Pclass:")
display(survival_by_pclass)


# 3. Average Age and Fare by Pclass
average_age_fare_by_pclass = (
    titanic_df.groupby("Pclass")[["Age", "Fare"]]
    .mean()
)

print("\n3. Average Age and Fare by Pclass:")
display(average_age_fare_by_pclass)


# 4. Passenger count and survival rate by Sex-Pclass
sex_pclass_summary = (
    titanic_df.groupby(["Sex", "Pclass"])
    .agg(
        Passenger_Count=("PassengerId", "count"),
        Survival_Rate=("Survived", "mean")
    )
)

sex_pclass_summary["Survival_Rate"] = (
    sex_pclass_summary["Survival_Rate"] * 100
)

print("\n4. Passenger count and survival rate by Sex-Pclass:")
display(sex_pclass_summary)


# 5. Passenger count, average Fare, and survival rate by Embarked
embarked_summary = (
    titanic_df.groupby("Embarked")
    .agg(
        Passenger_Count=("PassengerId", "count"),
        Average_Fare=("Fare", "mean"),
        Survival_Rate=("Survived", "mean")
    )
)

embarked_summary["Survival_Rate"] = (
    embarked_summary["Survival_Rate"] * 100
)

print("\n5. Passenger count, average Fare, and survival rate by Embarked:")
display(embarked_summary)
```

## Task 7 - Missing Values and Fare Outliers

```python
# -----------------------------
# Task 7: Missing Values and Fare Outliers
# -----------------------------

# 1. Missing-value count and percentage for every column

missing_count = titanic_df.isnull().sum()
missing_percentage = titanic_df.isnull().mean() * 100

missing_summary = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})

print("Missing values by column:")
display(missing_summary)


# 2. Plot missing-value counts

plt.figure(figsize=(10, 5))

missing_count.plot(kind="bar")

plt.title("Missing Values by Column")
plt.xlabel("Column")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)

plt.show()


# 3. Fill missing Age values with the mean Age

age_missing_before = titanic_df["Age"].isnull().sum()
age_mean = titanic_df["Age"].mean()

print("Missing Age values before mean imputation:", age_missing_before)
print("Mean Age:", age_mean)

titanic_df["Age"] = titanic_df["Age"].fillna(age_mean)

age_missing_after = titanic_df["Age"].isnull().sum()

print("Missing Age values after mean imputation:", age_missing_after)


# 4. Try different Age imputation methods

# Mean imputation
df_mean = titanic_df.copy()
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())

# Median imputation
df_median = titanic_df.copy()
df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())

# Mode imputation
df_mode = titanic_df.copy()
age_mode = df_mode["Age"].mode()[0]
df_mode["Age"] = df_mode["Age"].fillna(age_mode)

# Random-value imputation
df_random = titanic_df.copy()

missing_age_count = df_random["Age"].isnull().sum()

random_age_values = np.random.choice(
    df_random["Age"].dropna(),
    size=missing_age_count,
    replace=True
)

df_random.loc[df_random["Age"].isnull(), "Age"] = random_age_values


print("\nMissing Age values after each imputation method:")
print("Mean:", df_mean["Age"].isnull().sum())
print("Median:", df_median["Age"].isnull().sum())
print("Mode:", df_mode["Age"].isnull().sum())
print("Random:", df_random["Age"].isnull().sum())


# 5. Fare outlier detection using IQR

Q1 = titanic_df["Fare"].quantile(0.25)
Q3 = titanic_df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

fare_outliers = titanic_df[
    (titanic_df["Fare"] < lower_bound) |
    (titanic_df["Fare"] > upper_bound)
]

outlier_count = len(fare_outliers)

print("\nFare Outlier Analysis:")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("1.5 × IQR:", 1.5 * IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of Fare outliers:", outlier_count)
```

## Task 8 - Features and Pivot Table

```python
# -----------------------------
# Task 8: Features and Pivot Table
# -----------------------------

# 1. Create FamilySize
titanic_df["FamilySize"] = (
    titanic_df["SibSp"] +
    titanic_df["Parch"] +
    1
)

# 2. Create IsAlone
titanic_df["IsAlone"] = np.where(
    titanic_df["FamilySize"] == 1,
    1,
    0
)

print("New features:")
display(titanic_df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head())


# 3. Create pivot table
survival_pivot = pd.pivot_table(
    titanic_df,
    index="Sex",
    columns="Pclass",
    values="Survived",
    aggfunc="mean"
)

print("\nSurvival rate by Sex and Pclass:")
display(survival_pivot)


# 4. Find highest and lowest survival-rate groups
stacked_pivot = survival_pivot.stack()

highest_group = stacked_pivot.idxmax()
highest_survival_rate = stacked_pivot.max() * 100

lowest_group = stacked_pivot.idxmin()
lowest_survival_rate = stacked_pivot.min() * 100

print("\nHighest survival-rate group:")
print("Sex:", highest_group[0])
print("Pclass:", highest_group[1])
print("Survival rate:", highest_survival_rate, "%")

print("\nLowest survival-rate group:")
print("Sex:", lowest_group[0])
print("Pclass:", lowest_group[1])
print("Survival rate:", lowest_survival_rate, "%")
```

## Task 9 - Visualizations and Observations

```python
# -----------------------------
# Task 9: Visualizations and Observations
# -----------------------------

# 1. Correlation heatmap

relevant_numeric_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

correlation_matrix = titanic_df[relevant_numeric_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Titanic Numerical Variables")
plt.show()


# Find strongest positive and negative relationships
correlation_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

correlation_pairs = correlation_pairs.stack()

strongest_positive = correlation_pairs.idxmax()
strongest_positive_value = correlation_pairs.max()

strongest_negative = correlation_pairs.idxmin()
strongest_negative_value = correlation_pairs.min()

print("Strongest positive relationship:")
print(
    strongest_positive,
    "=", strongest_positive_value
)

print("\nStrongest negative relationship:")
print(
    strongest_negative,
    "=", strongest_negative_value
)


# 2. Survival rate by Sex

survival_rates_sex = titanic_df.groupby("Sex")["Survived"].mean() * 100

plt.figure(figsize=(7, 5))

survival_rates_sex.plot(kind="bar")

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)

plt.show()

print("\nSurvival rates by Sex:")
print(survival_rates_sex)


# 3. Age vs Fare, distinguished by survival

plt.figure(figsize=(9, 6))

survived_0 = titanic_df[titanic_df["Survived"] == 0]
survived_1 = titanic_df[titanic_df["Survived"] == 1]

plt.scatter(
    survived_0["Age"],
    survived_0["Fare"],
    label="Did not survive",
    alpha=0.6
)

plt.scatter(
    survived_1["Age"],
    survived_1["Fare"],
    label="Survived",
    alpha=0.6
)

plt.title("Age vs Fare by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend()

plt.show()
```
